**Step 1: Import the dependencies**

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report
```

**Step 2: Data preprocessing**

```python
df = pd.read_csv('/content/student-lifestyle-and-stress-dataset.csv', encoding = 'latin-1')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Stress_Level'].value_counts()
```

```python
X = df.iloc[:, 0:8]
```

```python
y = df.iloc[:, 8]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, stratify = y, random_state = 11)
```

```python
cat_cols = ['Student_Type']
```

```python
num_cols = ['Study_Hours', 'Social_Media_Hours', 'Attendance', 'Exam_Pressure', 'Family_Support', 'Month']
```

```python
cat_pipeline = Pipeline(steps = [
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('one-hot-encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=True))
])
```

```python
num_pipeline = Pipeline(steps = [
    ('impute', SimpleImputer(strategy='mean')),
    ('scale', StandardScaler())
])
```

```python
col_transformer = ColumnTransformer(transformers = [
    ('cat_pipeline', cat_pipeline, cat_cols),
    ('num_pipeline', num_pipeline, num_cols)],
    remainder = 'drop',
    n_jobs = -1)
```

**Step 3: Training model**

```python
model = RandomForestClassifier(class_weight='balanced',
                               random_state=42,
                               n_jobs=-1,
                               n_estimators=200,
                               max_depth=15,
                               min_samples_split=10,
                               max_features='sqrt')
```

```python
pipefinal = make_pipeline(col_transformer, model)
```

```python
pipefinal.fit(X_train, y_train)
```

```python
pipefinal.score(X_test, y_test)
```

```python
pipefinal.score(X_train, y_train)
```

```python
y_pred = pipefinal.predict(X_test)
```

**Step 4: Model evaluation**

```python
print(confusion_matrix(y_pred, y_test))
```

```python
print(classification_report(y_pred, y_test))
```
